## Importing Library

```python
import pandas as pd
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, models
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
import joblib
```

## Importing Dataset

```python

df = pd.read_csv('cleaned_riasec.csv')
# Thay thế 2 dòng bị lỗi bằng 2 dòng này:
features = ['R_Score', 'I_Score', 'A_Score', 'S_Score', 'E_Score', 'C_Score', 'TIPI1', 'TIPI2', 'TIPI3', 'TIPI4', 'TIPI5', 'TIPI6', 'TIPI7', 'TIPI8', 'TIPI9', 'TIPI10']
target_col = 'major'
X = df[features].values
y = df[target_col].values
```

## Label Encoder

```python
label_encoder = LabelEncoder()
y_encoded = label_encoder.fit_transform(y)
num_classes = len(label_encoder.classes_)
```

## Splitting Data

```python
X_train, X_test, y_train, y_test = train_test_split(X, y_encoded, test_size=0.2, random_state=42)
```

## Feature Scaling

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

## Saving Scaler & Label Encoder

```python
# joblib.dump(scaler, 'scaler.pkl')
# joblib.dump(label_encoder, 'label_encoder.pkl')
# print(f"Đã xử lý xong dữ liệu cho {num_classes} ngành nghề!")
```

## Build MLP

```python
ann = tf.keras.models.Sequential()
ann.add(tf.keras.layers.Dense(units= 6, activation='relu'))
ann.add(tf.keras.layers.Dense(units= 6, activation='relu'))
ann.add(tf.keras.layers.Dense(num_classes, activation='softmax'))
# Biên dịch mô hình
ann.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy', # Dùng sparse vì nhãn y là dạng số nguyên
              metrics=['accuracy'])
```

## Training Model

```python
history = ann.fit(
    X_train_scaled, y_train,
    epochs=100, 
    batch_size=32,
    validation_data=(X_test_scaled, y_test)
)
```
